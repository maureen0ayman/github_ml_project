```python
import requests
import pandas as pd
```

```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

repositories = data["items"]

print("Number of repositories:", len(repositories))
```

```python
df = pd.DataFrame(repositories)

df.head()
```

```python
df = df[
    [
        "name",
        "owner",
        "language",
        "stargazers_count",
        "forks_count",
        "watchers_count",
        "open_issues_count",
        "created_at",
        "updated_at",
        "license"
    ]
]

df.head()
```

```python
df["owner"] = df["owner"].apply(
    lambda x: x.get("login") if isinstance(x, dict) else None
)

df["license"] = df["license"].apply(
    lambda x: x.get("name") if isinstance(x, dict) else None
)

df.head()
```

```python
df.isnull().sum()
```

```python
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
df["created_at"] = pd.to_datetime(df["created_at"]).dt.date
df["updated_at"] = pd.to_datetime(df["updated_at"]).dt.date
```

```python
df = df.rename(
    columns={
        "stargazers_count": "stars",
        "forks_count": "forks",
        "watchers_count": "watchers",
        "open_issues_count": "open_issues",
        "created_at": "created_date",
        "updated_at": "updated_date"
    }
)
```

```python
df.head()
```

```python
df.to_csv("github_projects.csv", index=False)

print("Dataset saved successfully!")
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("github_projects.db")
```

```python
df.to_sql("Repositories", conn, if_exists="replace", index=False)

print("Database and table created successfully!")
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
WHERE stars > 10000
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
WHERE name LIKE '%Machine%'
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, stars, language
FROM Repositories
WHERE stars > 1000
AND forks > 100
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, stars, language
FROM Repositories
WHERE language = 'Python'
OR NOT language = 'Unknown'
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, stars
FROM Repositories
ORDER BY stars DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
ORDER BY stars DESC
LIMIT 10
"""

top10 = pd.read_sql_query(query, conn)

top10
```

```python
query = """
SELECT COUNT(*) AS total_repositories
FROM Repositories
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT AVG(stars) AS average_stars
FROM Repositories
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC
"""

language_counts = pd.read_sql_query(query, conn)

language_counts
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top10["name"], top10["stars"])

plt.xlabel("Stars")
plt.ylabel("Repository")
plt.title("Top 10 Most Popular Machine Learning Repositories")

plt.gca().invert_yaxis()

plt.show()
```

```python
df["created_date"] = pd.to_datetime(df["created_date"])
```

```python
creation_trends = (
    df.groupby(df["created_date"].dt.year)
      .size()
      .reset_index(name="repository_count")
)

creation_trends
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    creation_trends["created_date"],
    creation_trends["repository_count"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Repositories")
plt.title("Machine Learning Repository Creation Trends")

plt.show()
```
